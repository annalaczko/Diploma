```python
import json

with open("file.json", "r") as f:
    data = json.load(f)
```

```python
type(data["shapes"])
```

```python
image_height = data["imageHeight"]
image_width = data["imageWidth"]
```

```python
file = f"1 "
points = data["shapes"][0]['points']

for point in points:
    x=point[0]
    y=point[1]
    normalized_x = x / image_width
    normalized_y = y / image_height
    file += f"{normalized_x} {normalized_y} "
```
