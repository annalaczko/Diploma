```python
from ultralytics import YOLO
import os
```

```python
# pretrained szegmentációs modellből indulunk (transfer learning / finetune)
model = YOLO("yolov8n-seg.pt")  # n/s/m/l/x méretek

model.train(
    data="../data/mask.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
 # GPU id, vagy "cpu"
)

model.save("../models/yolov8n-seg-finetuned.pt")
```

```python
results = model.predict("../data/raw/00_Selection ML/2012_Survey Gladiatorenschule/A14_02.jpg")
```
